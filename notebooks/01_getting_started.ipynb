# Getting Started with python_magnetrun

This notebook demonstrates basic usage of python_magnetrun for analyzing magnet run data.

```python
# Import required libraries
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Import python_magnetrun modules

print(f"Python version: {sys.version}")
print("python_magnetrun installed successfully")
```

## Explore Available Data

Let's see what data files are available in the workspace.

```python
# List data files
data_dir = Path('/workspace/data')
print("Available data files:")
for file in sorted(data_dir.glob('*')):
    print(f"  - {file.name} ({file.stat().st_size / 1024:.1f} KB)")
```

```python
# List server data (JSON files)
srvdata_dir = Path('/workspace/srvdata')
json_files = list(srvdata_dir.glob('*.json'))
print(f"\nFound {len(json_files)} JSON server data files")
print("\nFirst 10 files:")
for file in sorted(json_files)[:10]:
    print(f"  - {file.name}")
```

## Load and Explore Sample Data

Let's load one of the CSV data files and explore its structure.

```python
# Try loading a CSV file if available
csv_files = list(data_dir.glob('*.csv'))
if csv_files:
    sample_file = csv_files[0]
    print(f"Loading: {sample_file.name}")

    df = pd.read_csv(sample_file)
    print(f"\nDataFrame shape: {df.shape}")
    print(f"\nColumns: {df.columns.tolist()}")
    print("\nFirst few rows:")
    display(df.head())
else:
    print("No CSV files found in data directory")
```

## Example Visualization

If data is available, create a simple visualization.

```python
# Example plot (customize based on your data)
if csv_files and not df.empty:
    plt.figure(figsize=(12, 6))

    # Plot first numerical column if available
    numeric_cols = df.select_dtypes(include=[np.number]).columns
    if len(numeric_cols) > 0:
        df[numeric_cols[0]].plot()
        plt.title(f'{numeric_cols[0]} over time')
        plt.xlabel('Index')
        plt.ylabel(numeric_cols[0])
        plt.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
    else:
        print("No numeric columns found for plotting")
```
